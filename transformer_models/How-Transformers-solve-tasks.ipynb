### Transformer models for language

Language models are at the heart of modern NLP. They’re designed to understand and generate human language by learning the statistical patterns and relationships between words or tokens in text.

The Transformer was initially designed for machine translation, and since then, it has become the default architecture for solving all AI tasks. Some tasks lend themselves to the Transformer’s encoder structure, while others are better suited for the decoder. Still, other tasks make use of both the Transformer’s encoder-decoder structure.

### How language models work

Language models work by being trained to predict the probability of a word given the context of surrounding words. This gives them a foundational understanding of language that can generalize to other tasks.

There are two main approaches for training a transformer model:

- **Causal language modeling (CLM)**: The model is trained to predict the next word in a sequence, given the previous words. This is typically done using a decoder-only architecture, like GPT-2.

- **Masked language modeling (MLM)**: The model is trained to predict a masked word in a sequence, given the surrounding words. This is typically done using an encoder-only architecture, like BERT.


### Types of language models

In the Transformers library, language models generally fall into three architectural categories:

- **Encoder-only models**: These models, like `BERT`, are designed to understand the context of a given input. They are typically used for tasks like text classification, token classification, and question answering.

- **Decoder-only models**: These models, like `GPT-2`, `Llama`, are designed to generate text based on a given input. They are typically used for tasks like text generation and language modeling.

- **Encoder-decoder models**: These models, like `T5`, `BART`, combine both the encoder and decoder architectures. They are designed to take an input sequence and generate an output sequence, making them suitable for tasks like summarization and translation.

### Understanding the Transformer Architecture
it’s helpful to visualize how the **Encoder** and **Decoder** blocks interact. The core innovation of this architecture is the **Self-Attention** mechanism, which allows the model to weigh the importance of different words in a sequence regardless of their distance from each other.


### Key Components
*   **Multi-Head Attention:** This allows the model to jointly attend to information from different representation subspaces.
*   **Positional Encoding:** Since Transformers don't use `RNNs` or `CNNs`, they need a way to know the order of the words. This is added to the input embeddings.
*   **Feed-Forward Networks:** Each layer contains a fully connected feed-forward network applied to each position identically and independently.

The mathematical heart of the attention mechanism is defined by the formula:

$$ \text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V $$

Where:
*   $Q$ is the Query matrix.
*   $K$ is the Key matrix.
*   $V$ is the Value matrix.
*   $d_k$ is the dimension of the keys.

Understanding which part of the Transformer architecture (encoder, decoder, or both) is best suited for a particular NLP task is key to choosing the right model. Generally, tasks requiring `bidirectional context` use `encoders`, `tasks generating text` use `decoders`, and `tasks converting` one sequence to another use `encoder-decoders`.

#### Text generation

For text generation tasks, decoder-only models like `GPT-2` are typically used. These models are trained using causal language modeling (CLM), where the model learns to predict the next word in a sequence given the previous words. This allows them to generate coherent and contextually relevant text.

- `GPT-2` is a popular decoder-only model that has been widely used for text generation tasks. It has been trained on a large corpus of text and can generate high-quality text based on a given prompt. `GPT-2` uses `byte pair encoding (BPE)` to tokenize words and generate a token embedding. Positional encodings are added to the token embeddings to indicate the position of each token in the sequence. The input embeddings are passed through multiple decoder blocks to output some final hidden state. Within each decoder block, `GPT-2` uses a masked self-attention layer which means `GPT-2` can’t attend to future tokens. It is only allowed to attend to tokens on the left.

- The output from the decoder is passed to a language modeling head, which performs a linear transformation to convert the hidden states into logits. The label is the next token in the sequence, which are created by shifting the logits to the right by one. The cross-entropy loss is calculated between the shifted logits and the labels to output the next most likely token.

GPT-2’s pretraining objective is based entirely on `causal language modeling`, predicting the next word in a sequence. This makes GPT-2 especially good at tasks that involve generating text.

#### Text classification

For text classification tasks, encoder-only models like `BERT` are typically used. These models are trained using masked language modeling (MLM), where the model learns to predict a masked word in a sequence given the surrounding words. This allows them to understand the context of the input text.

Text classification involves assigning predefined categories to text documents, such as sentiment analysis, topic classification, or spam detection.

`BERT` is an encoder-only model and is the first model to effectively implement deep bidirectionality to learn richer representations of the text by attending to words on both sides.

1. `BERT` uses `WordPiece tokenization` to split words into subword units, which helps it handle out-of-vocabulary words. Each token is converted into an embedding, and positional encodings are added to these embeddings to provide information about the position of each token in the sequence.
2. `BERT` is pretrained with two objectives: masked language modeling and next-sentence prediction. For masked language modeling, BERT randomly masks some of the tokens in the input and trains the model to predict those masked tokens based on their context. For next-sentence prediction, BERT is trained to predict whether two sentences appear sequentially in the original text.
3. The input embeddings are passed through multiple encoder layers to output some final hidden states. The hidden state corresponding to the `[CLS]` token is typically used for classification tasks, as it is designed to capture the overall meaning of the input sequence. This hidden state is passed through a classification head, which consists of a linear layer that outputs logits corresponding to the predefined categories. The cross-entropy loss is calculated between the logits and the true labels to train the model.
